```python
import pandas as pd
import numpy as np
```

```python
# Series from list , object type is string 
country = ['India' , 'pakistan' , 'usa' , 'Nepal' , 'Srilanka']
pd.Series(country)
```

```python
# custom index
marks = [67,57,89,100]
subjects = ['maths' , 'english' , 'science' , 'hindi']

pd.Series(marks , index=subjects)
```

```python
# Series from dict
marks = {
    'maths' : 67,
    'english' : 57,
    'science' : 89,
    'hindi' : 100
}

marks_series = pd.Series(marks , name='kisi ke marks')
marks_series
```

#### Selecting rows from a Dataframe
- iloc - searches using index positions
- loc - searches using labels (if index is in string)

```python
type(movies.iloc[0])
```

```python
# multiple rows , fancy indexing also works
type(movies.iloc[0:5])
```

```python
type(movies.loc[0]) # pass strings here if index is string
```

```python
movies.loc[0:3 , 'title_x':'poster_path']
```

#### Filtering a dataframe

```python
mask = ipl['MatchNumber']== 'Final'
ipl[mask][['Season','WinningTeam']]
```

```python
ipl[ipl['SuperOver'] == 'Y'].shape[0]
```

```python
# Count of action movies with rating > 7.5
mask1 = movies['genres'].str.split('|').apply(lambda x: 'Action' in x)
mask2 = movies['imdb_rating'] >= 7.5

movies[mask1 & mask2].shape[0]
```

```python
ipl[(ipl['City'] == 'Kolkata') & (ipl['WinningTeam'] == 'Chennai Super Kings')].shape[0]
```

```python
# drop duplicates
temp = pd.Series([1,1,2,2,3,3,3,4,4])
temp.drop_duplicates(keep='last')
```

```python
temp.duplicated().sum() # duplicated or not
```

```python
temp = pd.Series([1,2,3,np.nan,5,6,np.nan,8,np.nan,10])
temp
```

```python
temp.size
```

```python
temp.count() # does not count null values
```

```python
temp.isnull().sum() # total missing values
```

```python
temp.dropna() # drops null values
```

```python
temp.fillna(temp.mean()) # missing value will be replaced by mean
```

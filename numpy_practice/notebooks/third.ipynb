Array functions

```python
a1 = np.random.random((3,3))
a2 = np.round(a1 * 100)
```

```python
np.prod(a2)
```

```python
# max/min/sum/prod
# 0 -> col and 1 -> row
np.max(a1 , axis=1)
```

```python
#mean/median/std/var
np.var(a1 , axis=1)
```

```python
# trigonometric functions
np.sin(a1)
```

```python
# dot product
a3 = np.arange(12).reshape(3,4)
a4 = np.arange(12 , 24).reshape(4,3)
np.dot(a2,a3)
```

```python
# log and exponents
np.log(a1)
np.exp(a1)
```

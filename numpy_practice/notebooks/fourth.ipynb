```python
import numpy as np
```

```python
a1=np.arange(12)
a2=np.arange(12).reshape(3,4)
a3=np.arange(27).reshape(3,3,3)
a3
```

##### Indexing and slicing

```python
a3[::2,0,::2]
```

```python
# index1 to index2(not included) : skip
a1[2:5:2]
```

##### Iterating

```python
for i in a2:
    print(i)
```

```python
a4 = np.arange(12).reshape(3,4)
# transpose
np.transpose(a4)
a4.T
```

```python
# ravel - converts n dimensions array into 1D array
np.ravel(a3)
```

#### Stacking & splitting

```python
a5 = np.arange(16).reshape(4,4)
```

```python
# Horizontal stacking
np.hstack((a5,a5,a5))
```

```python
# Vertical stacking
np.vstack((a5,a5,a5))
```

```python
# Vertical cut
np.hsplit(a2 , 4)
```

```python
# horizontal cut
np.vsplit(a2 , 3)
```
